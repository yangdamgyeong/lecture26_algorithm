```python
# 리스트와 반복구조
# 주어진 데이터가 어디에 저장되어 있는지 출력

data_list = [23, 45, 21, 14, 2, 67]
target = 14

index = 0
while index <= 5:
    if data_list[index] == target:
        break
    index += 1

print(index)
```

```python
for index in range(6):
    if data_list[index] == target:
        break

print(index)
```

```python
# 리스트에 저장된 숫자 중 가장 큰 숫자 출력(최대값)

data_list = [23, 45, 21, 14, 2, 67]

max = data_list[0]
for index in range(1, 6):
    if data_list[index] > max:
        max = data_list[index]

print(max)
```

```python
data_list = [23, 45, 21, 14, 2, 67]

max = data_list[0]
for data in data_list:
    if data > max:
        max = data

print(max)
```

```python
# 리스트에 저장된 숫자 중 가장 작은 숫자 출력(최소값)

data_list = [23, 45, 21, 14, 2, 67]

min = data_list[0]
for index in range(1, 6):
    if data_list[index] < min:
        min = data_list[index]

print(min)
```

```python
# 리스트에 저장된 숫자 중 입력된 숫자보다 큰 수가 몇 개 있는지 구하기

data_list = [23, 45, 21, 14, 2, 67]

target = int(input('숫자 입력해주세요: '))

count = 0
for index in range(0, 6):
    if data_list[index] > target:
        count += 1

print(count)
```

```python
#  리스트에 저장된 모든 값 더하여 출력(합)

data_list = [23, 45, 21, 14, 2, 67]

sum = 0
for index in range(0, 6):
    sum += data_list[index]

print(sum)
```

```python
#  리스트에 저장된 숫자를 역순으로 저장하여 출력

data_list = [23, 45, 21, 14, 2, 67]

j = 5

for index in range(0, 3):
    temp = data_list[i]
    data_list[index] = data_list[j]
    data_list[j] = temp

    j -= 1
    
print(data_list)
```
