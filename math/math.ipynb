```python
# 최대 공약수

def getGCD(M, N):
    if M > N: M, N = N, M # M이 작은 숫자가 되도록 조정

    for num in range(M, 0, -1):
        if M % num == 0 and N % num == 0:
            return num

print(getGCD(12, 18))
```

```python
# 최소공배수

def getLCM(M, N):
    if M > N: M, N = N, M # M이 작은 숫자가 되도록 조정

    # 큰 숫자 N의 배수를 구하면서, M의 배수인지 확인
    # 최소공배수 -> 공배수 중 가장 작은 애 -> 처음 만나는 애
    i = 1
    while True:
        multiple = N * i
        if multiple % M == 0: # M의 배수인지 확인
            return multiple
        i += 1
print(getLCM(3, 4))        
```

```python
def getGCD(M, N):
    #반복문으로 최대공약수(GCD) 계산
    if M > N:
        M, N = N, M  # M이 작은 수가 되도록

    while N % M != 0:
        M, N = N % M, M
    return M

def getLCM(M, N):
   #GCD 기반으로 최소공배수(LCM) 계산
    gcd = getGCD(M, N)
    return (M * N) // gcd


A, B = map(int, input().split())

gcd = getGCD(A, B)
lcm = getLCM(A, B)

print(gcd)
print(lcm)
```
